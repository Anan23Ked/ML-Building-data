## 1.2  2. Theory

### In the Subtask A, the performance of algorithm Logistic Regression, Random Forest and Support Vector Machine (SVM) will be compared amongst each other in terms of the prediction of Gas Consumed or not in the building number 2 data values. Here, the theory behind these algorthims is described as such:
- Logistic Regression (Baseline): Binary classification, here there are two classes, if gas is consumed or not (0/1). This is a natural baseline for Binary Classificarion. It can be treated as a benchmark or a baseline algorithm against which others are compared.
- Random Forest: This used multiple decision Trees to make better predictions. This can handle non-linear interactions between features. For example the cold outdoor temperature AND night-time hour together predict gas use better than either alone
- SVM: This uses a different approach, maximising the distance between the classes, by providing the hyperpane that differentiates between the classes.

- Linear Regression (Baseline): This is a fundamental supervised learning algorithm used to model the relationship between a dependent variable (half hourly gam consupmtion) and one or more independent variables (temperature, humidity,. It predicts continuous values by fitting a straight line that best represents the data.
- Decision Tree: 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, ConfusionMatrixDisplay
from sklearn.preprocessing import StandardScaler
import time
import numpy as np
```

### Load and extract the relavent data. Building02 data

```python
data = './data/REFIT_TIME_SERIES_VALUES.csv'
df = pd.read_csv(data)
# print(df.head())
```

```python
def extract_data(var_id, col_name):
    some_values = [var_id]
    df_sel = df.loc[df['TimeSeriesVariable/@id'].isin(some_values)].copy()

    format = '%Y-%m-%dT%H:%M:%SZ'
    df_sel['dateTime_ix'] = pd.to_datetime(df_sel['dateTime'], format=format)
    df_sel = df_sel.set_index(pd.DatetimeIndex(df_sel['dateTime_ix']))

    df_sel = df_sel[['data']]
    df_sel.rename(columns={'data': col_name}, inplace=True)
    return df_sel
```

```python
# Extract Building 2 variables 
# Indoor temperature (one representative sensor per room)
df_int_temp  = extract_data('TimeSeriesVariable208',  'Int_Temp')

# Outdoor climate
df_ext_temp  = extract_data('TimeSeriesVariable1573', 'Ext_Temp')
df_humidity  = extract_data('TimeSeriesVariable1574', 'Humidity')
df_solar     = extract_data('TimeSeriesVariable1578', 'Solar')
df_wind      = extract_data('TimeSeriesVariable1575', 'Wind')

# Gas — Building 2 target variable
df_gas       = extract_data('TimeSeriesVariable1555', 'Gas')
# print(df_gas.head())
```

```python
# Feature selection 
FEATURES = ['Int_Temp', 'Ext_Temp', 'Humidity', 'Solar', 'Wind']
TARGET   = 'Gas'

df_merge['Hour']      = df_merge.index.hour
df_merge['Month']     = df_merge.index.month
df_merge['DayOfWeek'] = df_merge.index.dayofweek
FEATURES += ['Hour', 'Month', 'DayOfWeek']

df_model = df_merge[FEATURES + [TARGET]].dropna()
print(f"Modelling dataset shape: {df_model.shape}")
```

```python
# Train / val / test split
n         = len(df_model)
train_end = int(n * 0.70)
val_end   = int(n * 0.85)

df_train = df_model.iloc[:train_end]
df_val   = df_model.iloc[train_end:val_end]
df_test  = df_model.iloc[val_end:]

print(f"Train: {len(df_train):,} rows")
print(f"Validation: {len(df_val):,} rows")
print(f"Test: {len(df_test):,} rows")

X_train = df_train[FEATURES]
X_val   = df_val[FEATURES]
X_test  = df_test[FEATURES]

# Subtask A — binary classification target
y_train_cls = (df_train[TARGET] > 0).astype(int)
y_val_cls   = (df_val[TARGET]   > 0).astype(int)
y_test_cls  = (df_test[TARGET]  > 0).astype(int)

# Subtask B — regression target
y_train_reg = df_train[TARGET]
y_val_reg   = df_val[TARGET]
y_test_reg  = df_test[TARGET]

# Scale for SVM / Logistic Regression
scaler      = StandardScaler()
X_train_sc  = scaler.fit_transform(X_train)
X_val_sc    = scaler.transform(X_val)
X_test_sc   = scaler.transform(X_test)
```

## Subtask A - Classification

### A classification model is a type of machine learning model that sorts data points into predefined groups called classes. Classifiers learn class characteristics from input data, then learn to assign possible classes to new unseen data according to those learned characteristics

```python
# SUBTASK A — CLASSIFICATION
results_cls = []

def run_classifier(name, model, X_tr, y_tr, X_v, y_v):
    t0 = time.time()
    model.fit(X_tr, y_tr)
    train_time = time.time() - t0

    t0 = time.time()
    y_pred = model.predict(X_v)
    val_time = time.time() - t0

    tn, fp, fn, tp = confusion_matrix(y_v, y_pred).ravel()
    acc = accuracy_score(y_v, y_pred)
    f1  = f1_score(y_v, y_pred)

    results_cls.append({
        'Algorithm' : name,
        'Accuracy'  : round(acc, 4),
        'F1'        : round(f1,  4),
        'TP'        : tp, 'TN': tn, 'FP': fp, 'FN': fn,
        'Train time': round(train_time, 4),
        'Val time'  : round(val_time,   4),
        'y_pred'    : y_pred,
        'model'     : model,
    })
run_classifier('Logistic Regression',
               LogisticRegression(max_iter=1000, random_state=42),
               X_train_sc, y_train_cls, X_val_sc, y_val_cls)

run_classifier('Random Forest',
               RandomForestClassifier(n_estimators=100, random_state=42),
               X_train, y_train_cls, X_val, y_val_cls)

run_classifier('SVM',
               SVC(kernel='rbf', C=1, random_state=42),
               X_train_sc, y_train_cls, X_val_sc, y_val_cls)

df_results = pd.DataFrame(results_cls)
print(df_results[['Algorithm', 'Accuracy', 'F1', 'TP', 'TN', 'FP', 'FN', 'Train time', 'Val time']].to_string(index=False))
```

```python
# Confusion matrices ───────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, r in zip(axes, results_cls):
    cm = confusion_matrix(y_val_cls, r['y_pred'])
    ConfusionMatrixDisplay(cm, display_labels=['No Gas','Gas']).plot(
        ax=ax, colorbar=False, cmap='Blues')
    ax.set_title(r['Algorithm'])
plt.suptitle("Subtask A — Confusion Matrices", y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

# --- Subtask A: Accuracy & F1 ---
algos_cls = [r['Algorithm'] for r in results_cls]
acc_vals  = [r['Accuracy']  for r in results_cls]
f1_vals   = [r['F1']        for r in results_cls]

x = np.arange(len(algos_cls))
w = 0.35

ax.bar(x - w/2, acc_vals, width=w, label='Accuracy')
ax.bar(x + w/2, f1_vals,  width=w, label='F1 Score')

ax.set_xticks(x)
ax.set_xticklabels(algos_cls, rotation=15, ha='right')
ax.set_ylim(0, 1)
ax.set_title('Subtask A — Accuracy & F1')
ax.set_ylabel('Score')
ax.legend()

plt.tight_layout()
plt.show()
```

## Subtask 2: Regression


### Random Forest (accuracy=0.786, F1=0.736) outperforms both Logistic Regression and SVM. It handles the non-linear relationship between temperature and gas usage better than the linear models. SVM has the fewest false positives (17) but misses more true gas events (FN=19), making it too conservative.

### Predict actual gas consumption in m³

```python
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
results_reg = []
def run_regressor(name, model, X_tr, y_tr, X_v, y_v):
    # Train
    t0 = time.time()
    model.fit(X_tr, y_tr)
    train_time = time.time() - t0

    # Validate
    t0 = time.time()
    y_pred = model.predict(X_v)
    val_time = time.time() - t0

    # Metrics
    rmse    = np.sqrt(mean_squared_error(y_v, y_pred))
    cv_rmse = rmse / y_v.mean()          # CV-RMSE = RMSE / mean of observed
    mae     = mean_absolute_error(y_v, y_pred)
    r2      = r2_score(y_v, y_pred)

    results_reg.append({
        'Algorithm' : name,
        'CV-RMSE'   : round(cv_rmse, 4),
        'RMSE'      : round(rmse,    4),
        'MAE'       : round(mae,     4),
        'R2'        : round(r2,      4),
        'Train time': round(train_time, 4),
        'Val time'  : round(val_time,   4),
        'y_pred'    : y_pred,
        'model'     : model,
    })
```

```python
# Running all regression models 
# Linear Regression
run_regressor('Linear Regression',
              LinearRegression(),
              X_train, y_train_reg, X_val, y_val_reg)

# Lasso — regularised linear
run_regressor('Lasso',
              Lasso(alpha=0.01, max_iter=5000),
              X_train_sc, y_train_reg, X_val_sc, y_val_reg)

# Decision Tree Regressor 
run_regressor('Decision Tree',
              DecisionTreeRegressor(max_depth=5, random_state=42),
              X_train, y_train_reg, X_val, y_val_reg)

# SVR 
run_regressor('SVR',
              SVR(kernel='rbf', C=1),
              X_train_sc, y_train_reg, X_val_sc, y_val_reg)

# MLP Neural Network
run_regressor('Neural Network (MLP)',
              MLPRegressor(hidden_layer_sizes=(64, 32),
                           max_iter=500, random_state=42),
              X_train_sc, y_train_reg, X_val_sc, y_val_reg)
```

```python
# Hyperparameter sensitivity — Decision Tree (max_depth) 
depth_values = [2, 3, 5, 8, 10, 15, 20]
dt_sens = []

for d in depth_values:
    m = DecisionTreeRegressor(max_depth=d, random_state=42)
    m.fit(X_train, y_train_reg)
    y_pred = m.predict(X_val)
    rmse = np.sqrt(mean_squared_error(y_val_reg, y_pred))
    r2   = r2_score(y_val_reg, y_pred)
    dt_sens.append({'max_depth': d, 'RMSE': rmse, 'R2': r2})
    print(f"max_depth={d:3d}  RMSE={rmse:.4f}  R2={r2:.4f}")

df_dt_sens = pd.DataFrame(dt_sens)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(df_dt_sens['max_depth'], df_dt_sens['RMSE'],
             marker='o', color='steelblue')
axes[0].set_title('Decision Tree — RMSE vs max_depth')
axes[0].set_xlabel('max_depth')
axes[0].set_ylabel('RMSE')

axes[1].plot(df_dt_sens['max_depth'], df_dt_sens['R2'],
             marker='o', color='darkorange')
axes[1].set_title('Decision Tree — R² vs max_depth')
axes[1].set_xlabel('max_depth')
axes[1].set_ylabel('R²')
plt.tight_layout()
plt.show()
```

```python
layer_sizes = [(32,), (64,), (128,), (64, 32), (128, 64), (128, 64, 32)]
mlp_sens = []

for size in layer_sizes:
    m = MLPRegressor(hidden_layer_sizes=size, max_iter=500, random_state=42)
    m.fit(X_train_sc, y_train_reg)
    y_pred = m.predict(X_val_sc)
    rmse = np.sqrt(mean_squared_error(y_val_reg, y_pred))
    r2   = r2_score(y_val_reg, y_pred)
    mlp_sens.append({'layers': str(size), 'RMSE': rmse, 'R2': r2})
    print(f"layers={str(size):20s}  RMSE={rmse:.4f}  R2={r2:.4f}")

df_mlp_sens = pd.DataFrame(mlp_sens)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].bar(df_mlp_sens['layers'], df_mlp_sens['RMSE'], color='steelblue')
axes[0].set_title('MLP — RMSE vs layer config')
axes[0].set_xlabel('Hidden layers')
axes[0].set_ylabel('RMSE')
axes[0].tick_params(axis='x', rotation=30)

axes[1].bar(df_mlp_sens['layers'], df_mlp_sens['R2'], color='darkorange')
axes[1].set_title('MLP — R² vs layer config')
axes[1].set_xlabel('Hidden layers')
axes[1].set_ylabel('R²')
axes[1].tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Predicted vs actual plots 
fig, axes = plt.subplots(1, len(results_reg), figsize=(18, 4))

for ax, r in zip(axes, results_reg):
    ax.scatter(y_val_reg, r['y_pred'], alpha=0.3, s=8, color='steelblue')
    # Perfect prediction line
    mn = min(y_val_reg.min(), r['y_pred'].min())
    mx = max(y_val_reg.max(), r['y_pred'].max())
    ax.plot([mn, mx], [mn, mx], 'r--', linewidth=1)
    ax.set_title(r['Algorithm'])
    ax.set_xlabel('Actual Gas (m³)')
    ax.set_ylabel('Predicted Gas (m³)')
    ax.annotate(f"R²={r['R2']}", xy=(0.05, 0.92),
                xycoords='axes fraction', fontsize=9)

plt.suptitle("Subtask B — Predicted vs Actual", y=1.02)
plt.tight_layout()
plt.show()
```

## Task 2: K-means Clustering

### Loading and extracting the same data as above

```python
temp_df = df[df['TimeSeriesVariable/@id'] == 'TimeSeriesVariable1573'][['dateTime','data']].rename(columns={'data':'temperature'})
hum_df  = df[df['TimeSeriesVariable/@id'] == 'TimeSeriesVariable1574'][['dateTime','data']].rename(columns={'data':'humidity'})

merged = pd.merge(temp_df, hum_df, on='dateTime').dropna()
merged['dateTime'] = pd.to_datetime(merged['dateTime'])
```

```python
print(df['TimeSeriesVariable/@id'].unique())
```

```python
merged = merged.set_index('dateTime').resample('1H').mean().dropna().reset_index()
```

```python
from sklearn.preprocessing import StandardScaler

X = merged[['temperature', 'humidity']].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K_range = range(1, 11)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8,4))
plt.plot(K_range, inertia, 'bo-')
plt.xlabel('Number of clusters k')
plt.ylabel('Squared Error (Inertia)')
plt.title('Elbow Method for Optimal k')
plt.grid(True)
plt.show()
```

```python
k = 2

km_final = KMeans(n_clusters=k, random_state=42, n_init=10)
merged['cluster'] = km_final.fit_predict(X_scaled)
centroids_scaled = km_final.cluster_centers_
centroids = scaler.inverse_transform(centroids_scaled)
print("Centroids (temperature, humidity):", centroids)
```

### The elbow plot shows that inertia decreases sharply from k=1 to k=2, after which the rate of reduction becomes marginal. Therefore k=2 was selected as the optimal number of clusters, as adding further clusters beyond this point yields diminishing returns in explained variance.

```python
plt.figure(figsize=(8,6))
scatter = plt.scatter(merged['temperature'], merged['humidity'],
                      c=merged['cluster'], cmap='Set1', alpha=0.5, s=10)
plt.scatter(centroids[:,0], centroids[:,1],
            marker='X', s=200, c='black', label='Centroids')
plt.xlabel('Temperature (°C)')
plt.ylabel('Relative Humidity (%)')
plt.title('K-Means Clustering of Thermal Conditions')
plt.legend()
plt.colorbar(scatter, label='Cluster')
plt.show()
```

### The thermal conditions within the monitored building can be grouped into three distinct clusters:
Cluster 1 (Cold & Dry): Low temperature and low humidity, 
   corresponding to winter months or unoccupied/unheated periods.

Cluster 2 (Warm & Humid): High temperature and high humidity, 
  likely representing summer or heavily occupied periods.

Cluster 3 (Mild & Transitional):Moderate temperature and humidity, 
  corresponding to spring/autumn or partially heated conditions.

The two features used are indoor air temperature (°C) and relative humidity (%), aggregated at hourly resolution. These were chosen because they directly characterise thermal comfort conditions in the building.
