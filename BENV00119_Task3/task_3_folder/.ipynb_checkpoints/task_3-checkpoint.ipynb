## 4. Action

For this task, actions that agent needs to choose to maximize the reward are defined in advance.

These are the ***four actions***:

action 0: battery_charge

action 1: battery_discharge

action 2: grid_import

action 3: grid_export

There are some limitations implemented to make the environment more realistic, for example, action 0 is banned when battery state of charge is 1 and action 1 is banned when battery state of charge is under 0.2.

They are mapped to control signals of the environment by following codes:

```python
def actions_agent(mg0, action):
    pv = mg0.pv
    load = mg0.load
    net_load = load - pv
            
    capa_to_charge = mg0.battery.capa_to_charge  # remaining capacity to charge of the battery
    p_charge_max = mg0.battery.p_charge_max  # charge speed of the battery
    p_charge = max(0, min(capa_to_charge, p_charge_max)) 
    # charge value for the time, here we ensure that the charging value would not exceed remaining capacity.
   
    capa_to_discharge = mg0.battery.capa_to_discharge  # capacity of discharge
    p_discharge_max = mg0.battery.p_discharge_max  # discharge speed of the battery
    p_discharge = max(0,min(net_load, capa_to_discharge, p_discharge_max))  
    # discharge value for the time, here we ensure that the discharging value would not exceed current capacity
    
    if action == 0:
        control_dict = {'pv_consummed': min(pv,load), # PV is set to be consumed to satisfy the building load in advance
                        'battery_charge': p_charge, # tell charge value to the environment 
                        'battery_discharge': 0,
                        'grid_import': 0,
                        'grid_export': max(0, pv - min(pv, load) - p_charge) 
    # if PV is affluent to both satisfy the building load and battery charge, the remaining will be exported to the grid
                            }                          
    elif action == 1:    
        control_dict = {'pv_consummed': min(pv,load),
                        'battery_charge': 0,
                        'battery_discharge': p_discharge,# tell discharge value to the environment 
                        'grid_import': max(0,load - min(pv,load) - p_discharge), 
    # if battery discharge cannot meet the building load, the remaining will be imported from the grid
                        'grid_export':0
                       }
    elif action == 2:
        control_dict = {'pv_consummed': min(pv,load), 
                        'battery_charge': 0,
                        'battery_discharge': 0,
                        'grid_import': abs(max(0, net_load)),
                        'grid_export':0
    # import net load from the gird, no action for battery 
                       }
    elif action == 3:
        control_dict = {'pv_consummed': min(pv,load),
                        'battery_charge': 0,
                        'battery_discharge': 0,
                        'grid_import': 0,
                        'grid_export':abs(min(net_load, 0))
    # export net load to the gird, no action for battery 
                       }
      
    return control_dict
   
    # the basic logic is that the building load is required to meet 
    # surplus PV power (i.e., net_load < 0) can be used to either export to the grid or charge the battery
    # if PV cannot meet the building load (i.e., net_load > 0), the option is discharging the battery or importing from the grid
```

## 5. Reward

In this, we are trying to reduce (minimize) the overall cost of the building. Therefore we can easily define the reward as the minus cost. ***So agent maximizing the reward means minimizing the cost.***

The primary cost in the operation of a microgrid is the electricity cost, which includes the cost of purchasing grid electricity. Battery cost involves the cycle cost of charging and discharging the battery, calculated through the battery life cycle to determine the cost per usage. Additionally, the cost calculation also considers environmental impacts, increasing costs through the penalty for CO<sub>2</sub> emissions. The sum of these costs are automatically calculated by the environment and are fed back at each time step based on the actions taken.


## 6. Initialising  Q-table

```python
def init_qtable(mg0, nb_action, net_load_params):

    state = [] # initialize an empty list to store states
    Q = {} # initialize an empty dictionary for the Q-table

    for i in range(round(net_load_params[0]),round(net_load_params[1])+1, 1):   
        # iterate over a range from the rounded lower bound of net_load to its rounded upper bound
        
        for j in np.arange(round(mg0.battery.soc_min,1),round(mg0.battery.soc_max+0.1,1),0.1):
        # iterate over a range from the rounded minimum state of charge (soc_min) of the battery to slightly above its maximum (soc_max)
            
            j = round(j,1)
            # round the current value of soc to one decimal place
            state.append((i,j)) 
            # append the tuple (i, j) to the state list, representing a state (battery SOC and net load)
    
    for s in state:

        Q[s] = {}
        # initialize an empty dictionary for each state in the Q-table
        
        for a in range(nb_action):
        # Iterate over the range of 4 possible actions 
            
            Q[s][a] = 0 # Initialize all elements in the Q-table at zero

    return Q
```

So based on the states and actions, we can get a peak at the dimensions of Q-table:

```python
num_net_load_states  = len(range(round(net_load_params[0]),round(net_load_params[1])+1, 1))
num_soc_states = len(np.arange(round(mg0.battery.soc_min,1),round(mg0.battery.soc_max+0.1,1),0.1))
total_states = num_net_load_states * num_soc_states
nb_action = 4

print(num_net_load_states, net_load_params[0], net_load_params[1], mg0.battery.soc_min, mg0.battery.soc_max, num_soc_states)
print("Q-table is: {} x {}".format(total_states, nb_action))
```

```python
def training_Q_Learning(mg, horizon, net_load_params, alpha, epsilon, gamma, nb_episode=100, param_label=""):
    
    nb_action = 4  # actions : battery charge, battery discharge, grid import, grid export
    Q = init_qtable(mg0, nb_action, net_load_params)        # hint: initialise a Q-table using the function of 'def init_qtable' 
    nb_state = len(Q)   # hint: number of states using the length of Q table
    
    # List to store total rewards per episode
    episode_rewards = []  

    #do NOT change
    print("\n")                                
    print(f"--- Training with Alpha={alpha}, Epsilon={epsilon}, Gamma={gamma} ---")
    print_welcome(0)
    for e in range(nb_episode + 1):
        # Print training progress
        value_print = f"\r{param_label} Training Progress: Episode {e}/{nb_episode}"
        sys.stdout.write(value_print)
        sys.stdout.flush()
    #do NOT change   
        
        # Start learning
        episode_cost = 0
        r = 0
        # Reset the environment
        mg.reset()  

        net_load = round(mg.load - mg.pv)  
        soc = round(mg.battery.soc, 1) 
        s = (net_load, soc) # states 
        
        for i in range(horizon):
            a = max_dict(Q[s])[0]  # action with max value for the given state 
            a, is_random = espilon_decreasing_greedy(a, epsilon, nb_action)  # action after epsilon greedy policy
            
            # Modify action based on battery state of charge
            if soc == 0.2 and a == 1: # do NOT change
                a = 2                 # do NOT change
            if soc == 1 and a == 0:   # do NOT change
                a = 1                 # do NOT change
            
            
            action = actions_agent(mg, a)  # Returns control dictionary for the action
            status = mg.run(action)  # Send the action to the environment
            r = -mg.get_cost()  # Get the cost at the current time
            episode_cost += mg.get_cost()
            
            # to set new state
            net_load = round(mg.load - mg.pv)
            soc = round(mg.battery.soc, 1)
            s_ = (net_load, soc)

            # Modify reward based on action
            if a == 0:                # do NOT change
               r += 35                # do NOT change
               episode_cost -= 35     # do NOT change
            
            # Update Q-table
            Q[s][a] = Q[s][a] + alpha*(r + gamma*max_dict(Q[s_])[1]- Q[s][a]) # hint: based on above Q-learning algorithm update equation
            s = s_       # hint: assign the new state

        # do NOT change
        episode_rewards.append(-episode_cost)  # Store negative of episode cost as reward
        epsilon = update_epsilon(epsilon)  # Decrease epsilon

    return Q, episode_rewards
```

```python
def testing_Q_Learning(mg, Q, horizon):
    mg.reset(testing=True)
    total_cost = 0
    costs = []  # List to store single-step info
    states = []
    actions = []  

    for i in range(horizon):
        net_load = round(mg.load - mg.pv) # mg.load and mg.pv can be used to get load and pv at a given time
        soc = round(mg.battery.soc, 1) # mg.battery.soc to get the charge of the battery
        s = (net_load, soc) # states combination of new states
        a = max_dict(Q[s])[0] # action with max value for the given state (refer helper_function.py for a hint)

        if soc == 0.2 and a == 1:  # do NOT change
            a = 2                  # do NOT change
        if soc == 1 and a == 0:    # do NOT change
            a = 1                  # do NOT change

        actions.append(a)  # Store the action
        action = actions_agent(mg, a)
        status = mg.run(action)
        cost = mg.get_cost()

        if a == 0:               # do NOT change
            cost -= 35           # do NOT change

        total_cost += cost
        costs.append(total_cost)
        states.append((int(net_load), round(mg.battery.soc, 1))) # add two variables of the states here

        net_load = round(mg.load - mg.pv)
        soc = round(mg.battery.soc, 1)
        s_ = (net_load, soc)
        a_ = max_dict(Q[s_])[0]

        s, a = s_, a_

    return costs, states, actions
```

### *2. Please explain why SARSA is an on-policy learning and Q-learning is an off-policy learning*

Answer:  SARSA is an on-policy learning because it updates using the action based on the nest action taken based on the same policy,. It is called on-policy learning because th policy being evaluated and tested are the same. 
Equation: Q[s][a] += alpha*(r + gamma*Q[s'][a'] - Q[s][a]), where a' follows the same epsilon-greedy policy.

Q-learning updates the by checking for local maxima regardless of what action is taken. It evaluates the greedy policy while following exploratory plocy, hence it is a off-policy learning. 
Equation: Q[s][a] += alpha * (r + gamma * max_dict(Q[s_])[1] - Q[s][a])  updates using max Q[s']
